```python
import pandas as pd

STOCK_DATA = "stock_data_cleaned.csv"
FACTOR_DATA = "factor_data_cleaned.csv"

stock_df = pd.read_csv(STOCK_DATA)
factor_df = pd.read_csv(FACTOR_DATA)
```

```python
stock_df["Dates"] = pd.to_datetime(stock_df["Dates"], format="%Y-%m-%d").dt.date
factor_df["Date"] = pd.to_datetime(factor_df["Date"], format="%Y-%m-%d").dt.date
```

```python
prices = stock_df.drop(columns="Dates")
returns = prices / prices.shift(1) - 1
returns
```

```python
returns["Dates"] = pd.to_datetime(stock_df["Dates"])
returns.info()
```

## Rolling windows

```python
from datetime import datetime

# start_date = datetime(day=1, month=1, year=2009)
# final_date = datetime(day=31, month=12, year=2022)
# formation_window_period = pd.Period("2009-01", freq="M")

# window_idx = 0
# windows = []

return_month = returns["Dates"].dt.to_period("M")

first_month = pd.Period("2009-01", freq="M")
last_month = return_month.max()

windows = []
step = 0

while True:
    formation_start = first_month + 3 * step
    formation_end = formation_start + 5  # six months, including the start month

    holding_start = formation_start + 6
    holding_end = formation_start + 8  # three months

    if holding_end > last_month:
        break

    windows.append({
        "formation_start": formation_start,
        "formation_end": formation_end,
        "holding_start": holding_start,
        "holding_end": holding_end,
    })

    step += 1

schedule = pd.DataFrame(windows)

print("Number of windows:", len(schedule))
print(schedule.head(1))
print(schedule.tail(1))
```
